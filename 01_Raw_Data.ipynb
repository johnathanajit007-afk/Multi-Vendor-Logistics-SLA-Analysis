```python
import numpy as np
import pandas as pd
import random
from datetime import datetime, timedelta

np.random.seed(42)
random.seed(42)
n_rows = 1000

vendors = ['Delhivery', 'BlueDart', 'EcomExpress']
vendor_col = [random.choice(vendors) for _ in range(n_rows)]

base_ids = [f"TRK{100000 + i}" for i in range(950)]
tracking_ids = base_ids + [random.choice(base_ids) for _ in range(50)]

zones = ['Metro', 'Tier 2', 'Tier 3']
zone_col = [random.choice(zones) for _ in range(n_rows)]

start_date = datetime(2026, 1, 1)
order_dates = [start_date + timedelta(days=random.randint(0, 60), hours=random.randint(8, 20)) for _ in range(n_rows)]

actual_days = []
for z in zone_col:
    base_sla = 2 if z == 'Metro' else (4 if z == 'Tier 2' else 6)
    delay = np.random.choice([0, 1, 2, 3, 5], p=[0.70, 0.12, 0.10, 0.05, 0.03])
    actual_days.append(base_sla + delay)

delivery_dates = [o + timedelta(days=int(d)) for o, d in zip(order_dates, actual_days)]

dirty_order_dates = [o.strftime("%Y-%m-%d") if random.random() > 0.15 else o.strftime("%d/%m/%Y") for o in order_dates]
dirty_delivery_dates = [d.strftime("%Y-%m-%d") if random.random() > 0.15 else d.strftime("%d/%m/%Y") for d in delivery_dates]

# Inject missing delivery dates (packages in transit)
for i in random.sample(range(n_rows), 40):
    dirty_delivery_dates[i] = None

df = pd.DataFrame({
    'Tracking_ID': tracking_ids,
    'Vendor_Name': vendor_col,
    'Destination_Zone': zone_col,
    'Dispatch_Date': dirty_order_dates,
    'Delivery_Date': dirty_delivery_dates,
    'Order_Value_INR': np.random.randint(500, 12000, size=n_rows)
})

df.to_csv("Raw_Logistics_Data.csv", index=False)
print("Saved Raw_Logistics_Data.csv successfully!")
```
